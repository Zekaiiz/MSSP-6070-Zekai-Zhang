```python
import pandas as pd
faang = pd.read_csv("/content/MSSP-6070-Zekai-Zhang/data/faang.csv")
faang['date'] = pd.to_datetime(faang['date'], errors='coerce' )
faang['volume'] = pd.to_numeric(faang['volume'], errors='coerce').fillna(0).astype(int)
faang = faang.sort_values(by=['date','ticker']).reset_index(drop=True)
```

```python
highest7 = faang.nlargest(7,'volume')
print(highest7)
```

```python
melted=pd.melt(faang,id_vars=['date','ticker'],value_vars=['open','high','low','close','volume'])
print(melted)
```
