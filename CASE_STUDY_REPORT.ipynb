2. Which groups/stratum performs best? Are there differences between groups/stratums?

3. How does each gender perform in the global assessment?

```python
gender_assessmnt = df.groupby('GENDER')[['MAT_S11','CR_S11','CC_S11','BIO_S11','ENG_S11',
                      'QR_PRO','CR_PRO','CC_PRO','WC_PRO','ENG_PRO']].median()

gender_assessmnt['global assessment'] = gender_assessmnt.mean(axis=1)

print(gender_assessmnt)
```
