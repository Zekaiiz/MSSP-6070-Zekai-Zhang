```python
from google.colab import files
uploaded = files.upload()
import pandas as pd
df = pd.read_csv('InternetSales.csv', encoding='latin1')
print(df.head())
```

```python
from sre_constants import error
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors = 'coerce' )
df['years'] = df['OrderDate'].dt.year
df['SalesAmount'] = pd.to_numeric(df['SalesAmount'], errors = 'coerce')
yearly_sales = df.groupby('years')['SalesAmount']. sum(). reset_index()
print(yearly_sales)
```

```python
import matplotlib.pyplot as plt
plt.bar(yearly_sales['years'], yearly_sales['SalesAmount'])
plt.xlabel('Years')
plt.ylabel('Sales Amount')
plt.title('Yearly Sales')
plt.savefig('yearly_sales.pdf', format='pdf')
from google.colab import files
files.download('yearly_sales.pdf')
plt.show()
```
